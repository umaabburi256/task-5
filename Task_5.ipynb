```python

# List operations
fruits = ["Apple", "Banana", "Mango"]

print("Original list:", fruits)

fruits.append("Orange")
print("After adding:", fruits)

fruits.remove("Banana")
print("After removing:", fruits)

fruits.sort()
print("Sorted list:", fruits)

print("Is Mango present?", "Mango" in fruits)
```

```python
# Tuple operations
numbers = (10, 20, 30, 40, 20)

print("Tuple:", numbers)
print("First element:", numbers[0])
print("Length:", len(numbers))
print("Count of 20:", numbers.count(20))
print("Index of 30:", numbers.index(30))
```

```python
# Set operations
a = {10, 20, 30, 40}
b = {30, 40, 50, 60}

print("Set A:", a)
print("Set B:", b)

a.add(70)
print("After adding:", a)

a.discard(10)
print("After removing:", a)

print("Union:", a.union(b))
print("Intersection:", a.intersection(b))
print("Difference:", a.difference(b))
```

```python
# Dictionary operations
student = {
    "name": "Ravi",
    "age": 20,
    "course": "Python"
}

print("Student details:", student)
print("Student name:", student["name"])

student["age"] = 21
student["city"] = "Hyderabad"

print("After updating:", student)

student.pop("course")
print("After removing course:", student)
```

```python
# Student dataset
students = [
    {"name": "Ravi", "marks": 85},
    {"name": "Sita", "marks": 92},
    {"name": "Arun", "marks": 76},
    {"name": "Priya", "marks": 88}
]

for student in students:
    print(student["name"], "-", student["marks"])

print("Students scoring above 80:")
for student in students:
    if student["marks"] > 80:
        print(student["name"])
```

```python
# Employee dataset
employees = {
    101: {"name": "Ramesh", "salary": 25000},
    102: {"name": "Priya", "salary": 30000},
    103: {"name": "Kiran", "salary": 28000}
}

# Display employees
for emp_id, details in employees.items():
    print(emp_id, details)

# Add a new employee
employees[104] = {
    "name": "Anu",
    "salary": 32000
}

# Update salary
employees[101]["salary"] = 27000

print("Updated employee dataset:")
print(employees)
```
